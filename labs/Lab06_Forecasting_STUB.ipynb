## Task 2 — Score both baselines at your horizon (30 minutes)

**Why baselines first?** A model's error means little on its own: is a miss of 600
taxis good or bad? You can only tell by putting it next to what a one-line rule
misses on the same hours. So you score two simple rules first, and the lower score is
the number your model must beat. In a real system the forecast would be made every
hour from live data. Here all of it comes from the saved eight weeks, so everyone
gets the same scores.

The two simple rules, called **baselines**:

- **Persistence:** the forecast for any hour is the value `H` hours earlier.
- **Seasonal-naive:** the forecast for any hour is the same hour a week earlier
  (168 hours back), whatever `H` is.

The number to beat is the lower of the two scores. Each score is an **MAE** (mean
absolute error): the average size of a miss, in taxis. Lower is better.

## Task 4 — Find where the model is wrong: the residuals (10 minutes)

**Why residuals?** One MAE says how big the misses are on average, not when they
happen. A **residual** is actual minus model, for one hour. Flat noise around zero
means the model has learned the daily and weekly cycle. **Structure in the residuals**
means a pattern: a particular day, or the same hours every day, where the residuals
sit clearly above or below zero. That is the feature your model is missing.

1. **Run the cell below.** It draws two charts: the residuals hour by hour
   (*Residuals: actual minus model*), then their average for each hour of the day
   (*Mean residual by hour of day*).
2. **Answer the two questions under the charts**, in the box below them.
