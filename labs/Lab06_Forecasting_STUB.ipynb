## Task 4 — Plot the residuals

1. **Run the cell below.** It subtracts your prediction from the actual and plots
   what is left (the *residuals*), then their average by hour of day.
2. **Look for** structure, as described under the charts, and write it in the box.

Flat noise means the cycle is fully captured. **Structure in the residuals** — a
particular weekday, a spike at a particular hour — is the feature you are missing.

````python
if filled(pred):
    common = s.index.intersection(pd.Series(pred).index)
    resid = s.loc[common] - pd.Series(pred).loc[common]
    show.series(resid, title="Residuals: actual minus model", ylabel="taxis", hline=0)
    show.profile(resid, by="hour", title="Mean residual by hour of day")
````

**What you should see:** residuals scattered around zero with no shape is the
good case. A hump at particular hours in the second chart means the model is
consistently wrong at those hours, and that is the feature it is missing. A residual
is actual minus model: a bar above zero means the model guessed **too low** at that
hour; a bar below zero means it guessed **too high**.
