### Compare that against what you wrote

Most people get the row count roughly right and both other answers wrong.

`isna()` reports **zero** blanks — yet the minimum humidity is **−999%**.

Humidity cannot be negative. −999 is this logger's way of writing *"the sensor did
not answer"*. Nobody told pandas that, so it treats −999 as an ordinary number: it
drags the average down, it survives every `dropna()` you will ever write, and it
will never appear in `isna()`.

**This is the single most useful thing you will learn today.** Real data contains
values that are not measurements, and they do not announce themselves.

### Now look at it

A table of numbers hides this. A picture does not. Run the cell below. **Look
for:** where the readings normally sit, and anything that is nowhere near them.

```python
show.series(df, col="humidity_pct", title="Humidity, every reading, all 60 days")
```

**What you should see:** a thin band of readings near the top of the chart,
between roughly 65 and 85 per cent, and dozens of vertical lines dropping to −999
at the bottom. Each line is a sensor failure. In `describe()` they were one number in a corner; here they are
impossible to miss. This is why every lab from now on draws the data before
touching it.

One more. The sensors were busy on 14 May. Run the cell and **look for** anything
that does not look like a measurement.

```python
three_days = df[(df.timestamp >= "2026-05-13") & (df.timestamp < "2026-05-16")]
show.series(three_days, col="temp_c", title="Temperature, 13 to 15 May",
            span=("2026-05-14 09:00", "2026-05-14 23:00"))
```

**What you should see:** a wobbling line, then a perfectly flat one for fourteen
hours in the shaded period, then wobbling again. A real room does not hold 23.87 °C
to two decimal places for fourteen hours. The sensor was stuck. Notice that no
value is out of range: a stuck sensor produces no outliers at all, only a flat line,
and only a picture shows it. You meet it again in Lab 5.
