## Predict: How bad will the haze get next week? (15 minutes)

*The Prediction League: one question per lab, settled by reality. Individual work: you enter on your own, under your own name.*

**The question:** The highest 24-hour PSI reading in any region of Singapore during the target week. The target is the week of Monday 2026-10-26.

**How it works:** you enter a short function, `predict(history)`, not a number. When
entries close, the league runs your function once on the latest data. When the target
is over, the real value is compared with yours. League page (with the closing time):

**Step 1.** **Run the cell below.** It prints the number from the simple rule,
*the highest reading in the last seven days*. That is the number to beat.

````python
fetch("sg_psi_hourly.csv")
psi = pd.read_csv("sg_psi_hourly.csv", parse_dates=["timestamp"])

import pandas as pd

def predict(history):
    # the baseline: the highest reading in the last seven days
    last_week = history[history["timestamp"] > history["timestamp"].max() - pd.Timedelta(days=7)]
    print("highest PSI each day:", last_week.groupby(last_week["timestamp"].dt.date)["psi_max"].max().tolist())
    return float(last_week["psi_max"].max())

print("The baseline predicts", round(predict(psi), 1), "PSI")
````

**What you should see:** *The baseline predicts 230.0 PSI*.

**How to beat the simple rule.** Your entry is a guess of next week's highest PSI. A good
guess has a small *miss*: the gap between your guess and what really happens (too high or
too low both count). **Average miss** is that gap averaged over many past weeks. An idea
beats the simple rule if its average miss is smaller.

**Worked example, the last three weeks.** The simple rule guesses that next week's highest
PSI equals the worst reading of the last 7 days:

1. Week of 28 Sep: guessed 132, real 167, missed by 35.
2. Week of 05 Oct: guessed 167, real 192, missed by 25.
3. Week of 12 Oct: guessed 192, real 230, missed by 38.

Average miss over these three weeks: 33 PSI. The league page shows the same numbers
when you test your entry.

**The six ideas.** Each was tried on all 88 past weeks. Lower average miss is
better. *Haze season* means only the September and October weeks.

1. **Simple rule**: the worst reading of the last 7 days. Average miss: 10 PSI over all weeks, 16 PSI in haze season.
2. **Blend**: half the simple rule, half the average of the last 4 weekly worsts. Average miss: 9 PSI over all weeks, 16 PSI in haze season.
3. **Four-week average**: the average of the last 4 weekly worsts. Average miss: 10 PSI over all weeks, 19 PSI in haze season.
4. **Trend**: the simple rule, scaled up if the last 3 days are worse than the 4 before, down if better. Average miss: 12 PSI over all weeks, 20 PSI in haze season.
5. **Last 3 days**: the worst reading of the last 3 days. Average miss: 10 PSI over all weeks, 21 PSI in haze season.
6. **Same week last year**: the worst reading in the same week a year ago. Average miss: 22 PSI over all weeks, 98 PSI in haze season.

**What this shows.**

1. The simple rule is hard to beat. Most clever ideas do worse.
2. Averages smooth out spikes, so during a haze episode they guess too low.
3. Last year is no guide: the worst week of autumn 2025 reached 72; autumn 2026 reached 230.
4. Haze comes from fires, wind and rain, which are not in the file. No rule built on the file alone sees a change coming.
