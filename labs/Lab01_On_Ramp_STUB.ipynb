### Compare that against what you wrote

**What you should see:** 17,354 rows, **zero** blank values in every column, and a
minimum humidity of **−999**. Most people get the row count roughly right and both
other answers wrong. The row count is 74 more than 12 × 24 × 60 = 17,280: the
logger wrote about 140 readings twice and missed 66 during an outage. You will
meet both in Lab 4.

`isna()` reports **zero** blanks — yet the minimum humidity is **−999%**.

Humidity cannot be negative. −999 is this logger's way of writing *"the sensor did
not answer"*. Nobody told pandas that, so it treats −999 as an ordinary number: it
drags the average down, it survives every `dropna()` you will ever write, and it
will never appear in `isna()`.

**This is the single most useful thing you will learn today.** Real data contains
values that are not measurements, and they do not announce themselves.
