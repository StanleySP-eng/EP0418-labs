```python
#@title ✏️ TYPE HERE: your name and admission number { run: "auto", display-mode: "form" }
# Seeing code here, not boxes? Type each answer between the quotes "" and press play.
#@markdown **Your full name**
Full_name = "" #@param {"type":"string","placeholder":"type your answer here"}
#@markdown **Your admission number**
Admission_number = "" #@param {"type":"string","placeholder":"type your answer here"}

def _lab1_answers():
    # What is typed in every answer box. In Colab it is read from the notebook itself, so
    # a box counts even if its play button was never pressed. Elsewhere: the values the
    # boxes set when they ran.
    import ast, re
    names = ["Full_name", "Admission_number", "Q16_number_of_rows", "Q16_first_date", "Q16_last_date", "Q16_cadence_in_minutes", "Q16_blanks_from_isna", "Fault_1_check", "Fault_1_column", "Fault_1_evidence", "Fault_2_check", "Fault_2_column", "Fault_2_evidence", "Fault_3_check", "Fault_3_column", "Fault_3_evidence", "Fault_4_check", "Fault_4_column", "Fault_4_evidence", "Fault_5_check", "Fault_5_column", "Fault_5_evidence", "Fault_6_check", "Fault_6_column", "Fault_6_evidence", "Q20_cadence_in_minutes", "Q20_times_equal_to_cadence", "Q20_times_in_all", "Q21_rows_in_raw", "Q21_rows_in_clean", "Q21_rows_in_grid", "Q21_rows_removed", "Q21_why_removed", "Q21_rows_added", "Q21_why_added", "Q22_detector", "Q22_column", "Q22_normal_number", "Q22_fault_number", "Q22_threshold"]
    ans = {k: globals().get(k) for k in names}
    try:
        from google.colab import _message
        nb = _message.blocking_request("get_ipynb", request="", timeout_sec=20)["ipynb"]
        pat = re.compile(r"^\s*(" + "|".join(names) + r")\s*=\s*(.+?)\s*#\s*@param")
        for cell in nb.get("cells", []):
            src = cell.get("source", "")
            for line in (src if isinstance(src, list) else src.splitlines()):
                m = pat.match(line)
                if not m:
                    continue
                try:
                    v = ast.literal_eval(m.group(2))
                except Exception:
                    continue
                if (str(v).strip() not in ("", "choose")
                        or str(ans.get(m.group(1)) or "").strip() in ("", "choose")):
                    ans[m.group(1)] = v
    except Exception:
        pass
    return ans

def _lab1_save():
    import json, time
    from IPython.display import display
    ans = _lab1_answers()
    ans["_saved_at"] = time.time()
    display({"text/plain": "LAB1-ANSWERS " + json.dumps(ans, ensure_ascii=False),
             "text/html": "<span style='color:#888;font-size:12px'>Answers saved.</span>"}, raw=True)
globals().get("_lab1_save", lambda: None)()   # keeps a hidden copy of every answer
```

```python
# ---- VERIFICATION for Question 21: press play. Do not type here, do not edit ----
_need = [n for n in ("raw", "clean", "grid") if n not in globals()]
if _need:
    print("Not ready yet. Still to make:", ", ".join(_need),
          "(Questions 18 to 20). Then run this cell again.")
else:
    def _lab1_verify():
        import re
        R = {"temp_c": (18, 35), "humidity_pct": (0, 100), "co2_ppm": (300, 5000),
             "occupancy": (0, float("inf")), "power_kw": (0, float("inf"))}
        S = list(R)
        def line(ok, what, why):
            print(("[PASS] " if ok else "[FAIL] ") + what)
            if not ok:
                print("       " + why)
        def table(d):
            # the time can be a column of any name, or the index
            is_time = pd.api.types.is_datetime64_any_dtype
            cols = [k for k in d.columns if is_time(d[k]) and d[k].notna().all()]
            if "timestamp" in cols:
                t = d["timestamp"]
            elif is_time(d.index):
                t = d.index.to_series()
            elif cols:
                t = d[cols[0]]
            else:
                t = pd.to_datetime(d["timestamp"])
            t = pd.Series(pd.DatetimeIndex(t).tz_localize(None))
            o = d[[k for k in S if k in d.columns]].reset_index(drop=True)
            o.insert(0, "timestamp", t)
            return o
        r = table(pd.read_csv(CSV, parse_dates=["timestamp"]))   # the file itself, not your raw
        c, g = table(clean), table(grid)
        print(f"raw: {len(r):,} rows   clean: {len(c):,} rows   grid: {len(g):,} rows")
        for name, d in (("clean", c), ("grid", g)):
            gone = [k for k in S if k not in d.columns]
            if gone:
                line(False, f"{name} has the timestamp and the five sensor columns",
                     "Keep the column names the same as in raw.")
                return
        same = not c["timestamp"].duplicated().any() and set(c["timestamp"]) == set(r["timestamp"])
        line(same, "clean has one row for every timestamp in raw",
             "A timestamp is repeated, or a whole timestamp has been lost.")
        line(c["timestamp"].is_monotonic_increasing, "clean is in time order", "Sort clean by timestamp.")
        ci = c.drop_duplicates("timestamp").set_index("timestamp")
        outside = filled = lost = changed = 0
        for k, (a, b) in R.items():
            rr = r.assign(v=r[k].where(r[k].between(a, b))).groupby("timestamp")["v"]
            lo, hi, n_good, n_all = rr.min(), rr.max(), rr.count(), rr.size()
            x = pd.to_numeric(ci[k], errors="coerce").reindex(lo.index)
            bad = ((x < a) | (x > b))
            outside += int(bad.sum())
            filled += int((x.notna() & ~bad & (n_good == 0)).sum())
            lost += int((x.isna() & (n_good == n_all)).sum())
            changed += int((x.notna() & ~bad & (n_good > 0) & ((x < lo - 1e-6) | (x > hi + 1e-6))).sum())
        line(outside == 0 and filled == 0, "clean has blanks where a reading was outside its range in Table 1",
             f"{outside} readings are still outside; {filled} were given a new number: make them blank.")
        line(lost == 0 and changed == 0, "clean kept the readings that were inside their range, unchanged",
             f"{lost} good readings are now blank or missing; {changed} have a different value.")
        step = r["timestamp"].drop_duplicates().sort_values().diff().mode().iloc[0]
        slots = pd.date_range(r["timestamp"].min(), r["timestamp"].max(), freq=step)
        line(len(g) == len(slots) and list(g["timestamp"]) == list(slots),
             "grid has one row per time slot at the usual time between readings",
             f"It has {len(g):,} rows. From the first reading to the last, in order, with no slot left out.")
        gi = g.drop_duplicates("timestamp").set_index("timestamp")
        empty = gi.index.difference(ci.index)
        gx, cx = gi.reindex(ci.index)[S].astype(float), ci[S].astype(float)
        wrong = int((gx.isna() != cx.isna()).sum().sum() + ((gx - cx).abs() > 1e-6).sum().sum())
        put_in = int(gi.loc[empty, S].notna().sum().sum())
        line(wrong == 0 and put_in == 0,
             "grid has the readings of clean, unchanged, and blanks where there is no reading",
             f"{wrong} readings are lost or different from clean; {put_in} empty places were filled in.")
        ans = globals().get("_lab1_answers", lambda: {})()
        def typed(k):
            return str(ans.get(k) or globals().get(k, "") or "").strip()
        def whole(k):
            m = re.search(r"[0-9][0-9 ,]*", typed(k))
            return int(re.sub("[ ,]", "", m.group(0))) if m else None
        want = [("rows in raw", "Q21_rows_in_raw", len(r)), ("rows in clean", "Q21_rows_in_clean", len(c)),
                ("rows in grid", "Q21_rows_in_grid", len(g)), ("rows removed", "Q21_rows_removed", len(r) - len(c)),
                ("rows added", "Q21_rows_added", len(g) - len(c))]
        off = [w for w, k, n in want if whole(k) != n]
        line(not off, "Question 21 box: your five numbers match these tables",
             "Empty or not matching: " + ", ".join(off) + ".")
        no_why = [w for w, k in (("why removed", "Q21_why_removed"), ("why added", "Q21_why_added"))
                  if len(re.findall(r"[A-Za-z]{2,}", typed(k))) < 2]
        line(not no_why, "Question 21 box: both reasons are typed",
             "Type a reason of at least two words: " + ", ".join(no_why) + ".")
        print("A marker reads your reasons and your code: PASS does not check them.")
    try:
        _lab1_verify()
    except Exception as _err:
        print("[FAIL] The check could not read clean or grid:", type(_err).__name__)
        print("       clean and grid must be tables with a timestamp column and the five sensor columns.")
```

```python
# ---- VERIFICATION for Question 23: press play. Do not type here, do not edit ----
if "detected" not in globals():
    print("Not ready yet. Accept and run Gemini's cell that makes `detected`,",
          "then run this cell again.")
else:
    try:
        def _lab1_times(d):
            # the times in detected: a column or the index, in any date format
            found = []
            for v in ([d.index] + [d[k] for k in d.columns] if isinstance(d, pd.DataFrame) else [d.index, d]):
                if isinstance(v, pd.PeriodIndex):
                    v = v.to_timestamp()
                if pd.api.types.is_numeric_dtype(v) or pd.api.types.is_bool_dtype(v):
                    continue
                t = pd.Series(pd.to_datetime(pd.Series(list(v)), errors="coerce"))
                if len(t) and t.notna().all():
                    found.append(t)
            return found[0] if found else None
        _n, _all = len(detected), len(globals().get("grid", globals().get("raw", [])))
        _flags = getattr(detected, "dtype", None) == bool
        _t = _lab1_times(detected) if isinstance(detected, (pd.DataFrame, pd.Series)) and _n and not _flags else None
        _days = _t is not None and bool((_t == _t.dt.normalize()).all())
        if _days and "grid" in globals():
            _gt = _lab1_times(grid)
            _all = _gt.dt.normalize().nunique() if _gt is not None else _all
        if not isinstance(detected, (pd.DataFrame, pd.Series)) or _flags:
            print("[FAIL] detected is a table of flagged rows")
            print("       Put only the flagged rows in a table called detected, not a list or True/False values.")
        elif _n == 0:
            print("[FAIL] detected has flagged rows")
            print("       Your detector flagged nothing. Check the column, the threshold and the direction.")
        elif _n > 0.6 * _all:
            print("[FAIL] detected has only the flagged rows")
            print(f"       It has {_n:,} of the {_all:,} {'days' if _days else 'rows'}: that is most of them.")
        else:
            print(f"detected has {_n:,} flagged {'days' if _days else 'rows'}.")
            if _t is None:
                print("No dates found: ask Gemini to keep the timestamp column in detected.")
            elif _days:
                print(f"Dates: {_t.min():%d %b} to {_t.max():%d %b}")
            else:
                _by = _t.groupby(_t.dt.normalize()).agg(["min", "max", "size"])
                for _day, _r in _by.head(8).iterrows():
                    print(f"Dates: {_day:%d %b}, {_r['min']:%H:%M} to {_r['max']:%H:%M} ({_r['size']} rows)")
                if len(_by) > 8:
                    print(f"Dates: and {len(_by) - 8} more, up to {_by.index[-1]:%d %b}")
            print("Compare them with the fault you chose. Nothing here says they are right.")
    except Exception as _err:
        print("[FAIL] The check could not read detected:", type(_err).__name__)
```
